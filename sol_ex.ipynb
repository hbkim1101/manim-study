```python
%%manim -ql -v WARNING Problem1

class Problem1(ProblemScene):
    def construct(self):
        self.create_layout_areas()
        self.wait(1)
        
        problem = pdf_problem_image("math_2709.pdf", problem_number=1)
        problem.scale_to_fit_width(self.problem_width)
        problem.move_to(self.problem_coord, aligned_edge=UL)
        
        t0, t1, t2, t3, t4, t5 = MathTex("=", "2", Fraction(3, 2), r"\times", "4", ["-", Fraction(1, 2)])
        eq1 = MathTex(Term(t1, t2), t3, Term(t4, t5))
        eq2 = MathTex(t0, Term(t1, t2), t3, Term(t4, t5)); eq2[3].base = change(eq1[2].base, Paren(Term(2, 2)))
        eq3 = MathTex(t0, Term(t1, t2), t3, Term(eq2[3].base.inner.base, [eq2[3].base.inner.exponent, r"\times", Paren(t5)]))
        eq4 = eq3.copy(); eq4[3].exponent = change(list(eq3[3].exponent), -1)
        eq5 = MathTex(t0, Term(change([eq4[1].base, eq4[3].base], 2), [eq4[1].exponent, eq4[3].exponent]))
        eq6 = eq5.copy(); eq6[1].exponent = change(eq5[1].exponent, Fraction(1, 2))
        
        solution = VGroup(eq1, eq2, eq3, eq4, eq5, eq6).arrange(DOWN, aligned_edge=LEFT, buff=MED_SMALL_BUFF)
        solution.move_to(
            self.solution_coord + 0.25 * DOWN,
            aligned_edge=UL,
        )
        
        self.add(problem)
        self.camera.frame.move_to(problem.get_center() + 0.25 * DOWN).scale(0.4)
        self.wait(1)
        
        self.play(self.camera.frame.animate.move_to(solution[0].get_center() + 0.25 * DOWN).scale(0.7))
        self.wait(1)
        
        self.add(solution[0])
        for current_solution, next_solution in zip(solution, solution[1:]):
            camera_shift = np.array([
                0,
                next_solution.get_center()[1] - current_solution.get_center()[1],
                0,
            ])
            self.play(
                TransformInStages.from_copy(current_solution, next_solution),
                self.camera.frame.animate.shift(camera_shift),
                lag_ratio=0.5,
            )
        self.wait(1)
        
        
        whole_scene = Group(problem, *solution)

        frame_width = self.camera.frame.get_width()
        frame_height = self.camera.frame.get_height()
        frame_aspect_ratio = frame_width / frame_height

        fit_width = max(
            whole_scene.width + 1,
            (whole_scene.height + 1) * frame_aspect_ratio,
        )

        self.play(
            self.camera.frame.animate
            .move_to(whole_scene.get_center())
            .scale(fit_width / frame_width)
        )
        self.wait(5)
```

```python
%%manim -ql -v WARNING Problem2

class Problem2(ProblemScene):
    def construct(self):
        self.create_layout_areas()
        self.wait(1)
        
        problem = pdf_problem_image("math_2709.pdf", problem_number=2)
        problem.scale_to_fit_width(self.problem_width)
        problem.move_to(self.problem_coord, aligned_edge=UL)
        
        t1, t2, t3, t4, t5, t6, t7, t8, t9, t10 = MathTex("f", "(", "x", ")", "=", "2", "x", "3", "-", "x")
        eq1 = MathTex(t1, t2, t3, t4, t5, t6, Term(t7, t8), t9, t10, "-4")
        eq2 = MathTex(r"\Rightarrow", t1, r"^{\prime}", t2, t3, t4, t5, change([t6, t8], 6), Term(t7, change(t8, 2)), t9, change(t10, 1))
        
        s0, s1, s2, s3, s4 = MathTex("=", "f", "(", ")", "2")
        eq3 = MathTex(r"\lim_{x \to 2}", Fraction([s1, s2, "x", s3, "-f(", s4, ")"], "x-2"))
        eq4 = MathTex(s0, s1, r"^{\prime}", s2, s4, s3)
        eq5 = MathTex(s0, eq2[7], eq2[8], eq2[9], eq2[10])
        

        solution = VGroup(eq1, eq2, eq3, eq4, eq5).arrange(DOWN, aligned_edge=LEFT, buff=MED_SMALL_BUFF)
        solution.move_to(
            self.solution_coord + 0.25 * DOWN,
            aligned_edge=UL,
        )
        
        self.add(problem)
        self.camera.frame.move_to(problem.get_center() + 0.25 * DOWN).scale(0.4)
        self.wait(1)
        
        self.play(self.camera.frame.animate.move_to(eq1.get_center() + 0.25 * DOWN).scale(0.7))
        self.wait(1)
        
        self.add(solution[0])
        for current_solution, next_solution in zip(solution[:-2], solution[1:]):
            camera_shift = [0, next_solution.get_center()[1] - current_solution.get_center()[1], 0]
            self.play(
                TransformInStages.from_copy(current_solution, next_solution),
                self.camera.frame.animate.shift(camera_shift),
                lag_ratio=0.5,
            )
        self.wait(1)
        self.play(self.camera.frame.animate.move_to(solution[1].get_center() + 0.25 * DOWN))
        self.wait(1)
        camera_shift = [0, solution[4].get_center()[1] - solution[1].get_center()[1], 0]
        self.play(
            TransformInStages.from_copy(solution[1], solution[4]),
            self.camera.frame.animate.shift(camera_shift)
        )
        self.wait(1)
        
        # eq4의 2를 복사해서 eq5의 x 자리에 넣음
        eq5[0].save_x()
        new_base = eq4[4].copy()
        eq5.terms[2].base = new_base
        eq5.insert(2, r"\cdot")
        eq5[2].set_color(BLACK)
        eq5[0].restore_x()

        # eq5의 변화 애니메이션 생성
        anim = TransformInStages.progress(eq5)

        # new_base가 갑자기 FadeIn되지 않고
        # 실제 eq4[4]에서 출발하도록 지정
        anim.intercept(new_base).set_source(eq4[4])

        self.play(anim)

        self.wait(1)
        
        eq5[0].save_x()
        eq5[0].save_y()
        eq5.terms = [s0, change([eq5[1], eq5[2], eq5[3], eq5[4], eq5[5]], 24)]
        eq5[0].restore_x()
        eq5[0].restore_y()
        
        self.play(TransformInStages.progress(eq5))
        self.wait(1)
        
        
        whole_scene = Group(problem, *solution)

        frame_width = self.camera.frame.get_width()
        frame_height = self.camera.frame.get_height()
        frame_aspect_ratio = frame_width / frame_height

        fit_width = max(
            whole_scene.width + 1,
            (whole_scene.height + 1) * frame_aspect_ratio,
        )

        self.play(
            self.camera.frame.animate
            .move_to(whole_scene.get_center())
            .scale(fit_width / frame_width)
        )
        self.wait(5)
```

```python
%%manim -ql -v WARNING Problem3

class Problem3(ProblemScene):
    def construct(self):
        self.create_layout_areas()
        self.wait(1)
        
        problem = pdf_problem_image("math_2709.pdf", problem_number=3)
        problem.scale_to_fit_width(self.problem_width)
        problem.move_to(self.problem_coord, aligned_edge=UL)
        
        self.add(problem)
        self.camera.frame.move_to(problem.get_center() + 0.25 * DOWN).scale(0.4)
        self.wait(1)
        
        eq1 = MathTex(r"a_{10}-a_7", "=", "9").move_to(self.solution_coord + 0.25 * DOWN, aligned_edge=UL)
        
        self.play(self.camera.frame.animate.move_to(eq1.get_center() + 0.25 * DOWN).scale(0.7))
        self.wait(1)
        
        self.add(eq1)
        eq1[0].save_x()
        eq1.insert(1,["=", 3, "d"])
        eq1.set_color(BLACK)
        eq1[0].restore_x()
        self.play(TransformInStages.progress(eq1))
        self.wait(1)
        
        eq2 = MathTex(eq1[1][2], eq1[2], Fraction(eq1[3], eq1[1][1])).next_to(eq1, DOWN, aligned_edge=LEFT, buff=MED_SMALL_BUFF)
        camera_shift = [0, eq2.get_center()[1] - eq1.get_center()[1], 0]
        self.play(
            TransformInStages.from_copy(eq1, eq2),
            self.camera.frame.animate.shift(camera_shift),
            lag_ratio=0.5,
        )
        self.wait(1)
        
        eq2[0].save_x()
        eq2[0].save_y()
        eq2.terms[2] = change(list(eq2[2]), 3)
        eq2[0].restore_x()
        eq2[0].restore_y()
        
        self.play(TransformInStages.progress(eq2))
        self.wait(1)
        
        eq3 = MathTex(r"a_7").next_to(eq2, DOWN, aligned_edge=LEFT, buff=MED_SMALL_BUFF)
        camera_shift = [0, eq3.get_center()[1] - eq2.get_center()[1], 0]
        self.play(
            FadeIn(eq3),
            self.camera.frame.animate.shift(camera_shift),
            
        )
        

        eq3[0].save_x()
        eq3[0].save_y()
        eq3.terms = [*eq3.terms, "=", r"a_2", "+", "5", "d"]
        eq3.set_color(BLACK)
        eq3[0].restore_x()
        eq3[0].restore_y()
        
        self.play(TransformInStages.progress(eq3))
        self.wait(1)
        
        eq4 = eq3.copy()
        eq4[2] = change(eq3[2], 2)
        eq4[5] = change(eq3[5], 3)
        eq4.insert(5,r"\cdot")
        eq4.set_color(BLACK).next_to(eq3, DOWN, aligned_edge=LEFT, buff=MED_SMALL_BUFF)
        
        camera_shift = [0, eq4.get_center()[1] - eq3.get_center()[1], 0]
        self.play(
            TransformInStages.from_copy(eq3, eq4),
            self.camera.frame.animate.shift(camera_shift),
            lag_ratio=0.5,
        )
        self.wait(1)
        
        eq4[0].save_x()
        eq4[0].save_y()
        eq4.terms = [eq4[0], eq4[1], change([eq4[2], eq4[3], eq4[4], eq4[5]], 17)]
        eq4.set_color(BLACK)
        eq4[0].restore_x()
        eq4[0].restore_y()

        self.play(
            TransformInStages.progress(eq4)
        )
        self.wait(1)
```
